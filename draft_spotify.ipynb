```python
import os
from dotenv import load_dotenv

import spotipy
from spotipy.oauth2 import SpotifyOAuth, SpotifyClientCredentials

load_dotenv()
```

```python
sp_user = spotipy.Spotify(
    auth_manager=SpotifyOAuth(scope="user-top-read")#"user-library-read")
)

sp_public = spotipy.Spotify(
    auth_manager=SpotifyClientCredentials(
        client_id=os.getenv("SPOTIPY_CLIENT_ID"),
        client_secret=os.getenv("SPOTIPY_CLIENT_SECRET")
    )
)
```

```python
results = sp_user.search(q="joy division", limit=20)
#for idx, track in enumerate(results["tracks"]["items"]):
#    print(idx, track["name"])
results
```

```python
track_id = results["tracks"]["items"][0]["id"]
result_analysis = sp_public.audio_features([track_id])
print(result_analysis)
```
